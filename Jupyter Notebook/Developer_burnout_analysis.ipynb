```python
# Numerical columns
num_cols = [ 
    'Upskilling hours per week',
    'Leetcode problems solved',
    'Certifications count',
    'Side projects count',
    'Github activity score',
    'Interview preparation hours weekly',
    'Caffeine intake per day',
    'Social media usage hours',
    'Emergency savings months'
]
```

```python
#  Fill with median
for columns in num_cols: df[columns] = df[columns].fillna(df[columns].median())
```

```python
# Categorical columns
cat_cols = [
    'Therapy or counseling',
    'Linkedin job search activity'
]
```

```python
# Fill categorical nulls with mode
for columns in cat_cols: df[columns] = df[columns].fillna(df[columns].mode()[0])
```

```python
# Check null values again
df.isnull().sum()
```

```python
# Convert column names to lowercase and replace spaces with underscores
df.columns = df.columns.str.lower().str.replace(' ','_')
```

```python
df.columns
```

```python
# Divide employees into experience groups
# create a column experience_group
num = [0, 2, 5, 10, 20, 30]
lables = ['Fresher', 'Early Career', 'Mid Level', 'Senior', 'Veteran']
```

```python
# Categorize employees into experience-based segments to simplify analysis and modeling
df['experience_group'] = pd.cut(
    df['experience_years'],
    bins=[0, 2, 5, 10, 20, 30],
    labels=['Fresher', 'Early Career', 'Mid Level', 'Senior', 'Veteran']
)
```

```python
# Standardize company size categories by fixing Excel-induced date format conversions
df.loc[df['company_size'].str.contains('Oct', na=False), 'company_size'] = '1-10'
df.loc[df['company_size'].str.contains('Nov', na=False), 'company_size'] = '11-50'
```
